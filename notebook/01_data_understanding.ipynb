```python
# Đếm số dòng có phí vận chuyển bằng 0.
zero_freight_count = order_items[
    order_items["freight_value"] == 0
].shape[0]

print("Số dòng có freight_value = 0:", zero_freight_count)
```

```python
zero_freight = order_items[
    order_items["freight_value"] == 0
]

zero_freight[
    ["order_id", "product_id", "price", "freight_value"]
].head(10)
```

```python
zero_freight["product_id"].value_counts().head(10)
```

```python
freight_analysis = (
    order_items.groupby("product_id").agg(
        total_orders = ("order_id", "count"),
        zero_freight_count = ("freight_value", lambda x: (x == 0).sum()),
    )
    .query("zero_freight_count > 0")
)

freight_analysis["zero_freight_percentage"] = (
    freight_analysis["zero_freight_count"] / freight_analysis["total_orders"] * 100
)

freight_analysis.sort_values("zero_freight_count", ascending=False)
```
